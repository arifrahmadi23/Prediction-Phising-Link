```python
BASE_PATH = "/content/drive/MyDrive/phishing_url/raw"

df_A = pd.read_csv(f"/content/drive/MyDrive/phishing_research_new/datasets/combined_dataset.csv")
df_B = pd.read_csv(f"/content/drive/MyDrive/phishing_research_new/datasets/dataset_link_phishing.csv")
df_C = pd.read_csv(f"/content/drive/MyDrive/phishing_research_new/datasets/url_dataset.csv")

print(df_A.shape, df_B.shape, df_C.shape)
```

```python
def standardize_dataset(df, url_col, label_col, phishing_label):
    df = df[[url_col, label_col]].copy()
    df.columns = ['url', 'label']

    df['label'] = df['label'].apply(
        lambda x: 1 if x == phishing_label else 0
    )

    df.dropna(inplace=True)
    df['url'] = df['url'].astype(str).str.lower()

    return df
```

```python
df_A = standardize_dataset(
    df_A,
    url_col="domain",
    label_col="label",
    phishing_label="1"
)

df_B = standardize_dataset(
    df_B,
    url_col="url",
    label_col="status",
    phishing_label="phishing"
)

df_C = standardize_dataset(
    df_C,
    url_col="url",
    label_col="type",
    phishing_label="phishing"
)
```

```python
df_all = pd.concat([df_A, df_B, df_C], axis=0)
df_all.reset_index(drop=True, inplace=True)

print("Total samples:", len(df_all))
print(df_all['label'].value_counts())
```

```python
SAVE_PATH = "/content/drive/MyDrive/phishing_ressearch_new/data_processed"

os.makedirs(SAVE_PATH, exist_ok=True)

df_all.to_csv(
    f"{SAVE_PATH}/phishing_url_integrated.csv",
    index=False
)

print("Dataset integrated saved successfully.")
```
